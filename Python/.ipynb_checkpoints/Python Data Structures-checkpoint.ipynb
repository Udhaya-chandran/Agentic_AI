- Mutable

## 2.Sets
- 1D
- Hetrogenous
- It won't allow duplicate values
- Indexing & Slicing is not possible
- Mutable

```python
empty_cart_2 = set()
```

```python
type(empty_cart_2)
```

```python
items_to_buy_2 = {"Apple", "Orange", "Grapes", "Pomgranate","Apple", 6.76,8,9}
items_to_buy_2
```

```python
type(items_to_buy_2)
```

```python
items_to_buy_2[1]
```

```python
items_to_buy_2[1]= "banana"
```

```python
items_to_buy_2.add("banana")
```

```python
items_to_buy_2
```

```python
items_to_buy_2.discard("Apple")
```

```python
items_to_buy_2
```

```python
items_to_buy_2.difference()
```

## 3. Tuple
- 1D
- Hetrogenous
- Duplicate Values is allowed
- Indexing & Slicing is possible
- Imutable

```python
empty_cart_3=()
```

```python
type(empty_cart_3)
```

```python
items_to_buy_3 = ("Apple", "Orange", "Grapes", "Pomgranate","Apple", 6.76,8,9)
items_to_buy_3
```

```python
items_to_buy_3[0:3]
```

```python
items_to_buy_3.count('Apple')
```

```python
items_to_buy_3.index('Grapes')
```

## 4. Dictionary

It is a Key:Value pair

- 2D
- Hetrogenous
- Mutable
- Indexing & Slicing is not possible
- Don't create a duplicate key, but you can make duplicate values

```python
empty_cart_4 = {}
empty_cart_4
```

```python
type(empty_cart_4)
```

```python
student_marks = {89,98,99}
```

```python
type(student_marks)
```

```python
student_marks = {"Maths" : 89 , "Tamil" : 100 , "English" : 98}
```

```python
type(student_marks)
```

```python
student_marks
```

```python
items_to_buy_4 = {"potato" : 5 , "tomato" : 2 , "Apple" : 3.5}
items_to_buy_4
```

```python
type(items_to_buy_4)
```

```python
items_to_buy_4[2]
```

```python
items_to_buy_4["Apple"]
```

```python
items_to_buy_4.clear()
```

```python
items_to_buy_4
```

```python
items_to_buy_4.keys()
```

```python
items_to_buy_4.values()
```
